```python
import numpy as np
from fugu import Scaffold
from fugu.bricks import Vector_Input, NoOps
from fugu.backends import snn_Backend
```

```python
NoOps.show_ports()
```

```python
noops_delay = 2

scaffold = Scaffold()
scaffold_input = scaffold.add_brick(Vector_Input(np.array([1]), name='input1'))
noops = scaffold.add_brick(NoOps(noops_delay=noops_delay), output=True)
scaffold.connect(scaffold_input, noops)
scaffold.lay_bricks()
scaffold.summary(verbose=1)

print("======================================")
backend = snn_Backend()
backend_args = {}
backend_args['record'] = 'all'
backend.compile(scaffold, backend_args)
result = backend.run(noops_delay + 5)
print(result)
```
